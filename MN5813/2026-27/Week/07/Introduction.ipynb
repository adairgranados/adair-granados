### Allocating audit slices

The cell below splits the two tables' columns into six slices: Four over the incidents table and two over the mobilisations table. Type your team's names into it and run it.

- Some members take a second slice. Second slices always come from the lightest pair, *when* and *what*
- Swap slices by agreement, and log the swap

````python
# --- Data setup: Run this cell first (runs anywhere, no download needed) ---
import numpy as np

team = ["Member 1", "Member 2", "Member 3", "Member 4"]  # Type your names here

slices = [
    "Incidents, where: Postcode_district, IncGeo_BoroughCode/Name, IncGeo_WardCode/Name, "
    "IncidentStationGround, AddressQualifier (and BoroughName, WardName in mobilisations)",
    "Incidents, response: FirstPumpArriving_AttendanceTime/DeployedFromStation, "
    "SecondPumpArriving_AttendanceTime/DeployedFromStation, NumStationsWithPumpsAttending, "
    "NumPumpsAttending, PumpCount, PumpMinutesRounded, Notional Cost (£), NumCalls",
    "Mobilisations, times: DateAndTimeMobilised, DateAndTimeMobile, DateAndTimeArrived, "
    "DateAndTimeLeft, DateAndTimeReturned, TurnoutTimeSeconds, TravelTimeSeconds, "
    "AttendanceTimeSeconds, PerformanceReporting",
    "Mobilisations, who and why: ResourceMobilisationId, Resource_Code, "
    "DeployedFromStation_Code/Name, DeployedFromLocation, PumpOrder, PlusCode_Code/Description, "
    "DelayCodeId, DelayCode_Description",
]
light = [
    "Incidents, when: DateOfCall, TimeOfCall, HourOfCall, CalYear "
    "(plus the join count on IncidentNumber, for everyone)",
    "Incidents, what: IncidentGroup, StopCodeDescription, SpecialServiceType, "
    "PropertyCategory, PropertyType (plus the duplicate-row count in both tables, for everyone)",
]  # Allocated last, so any second slice comes from this pair

rng = np.random.default_rng(5813)  # module seed: Do not change
order = list(rng.permutation(slices)) + list(rng.permutation(light))

print("Audit slice allocation:")
for i, s in enumerate(order):
    print(f"  {team[i % len(team)]}: {s}")
````

## 0:15–0:45: Data audit

**Individually (about 20 minutes):** Load the two tables with the `pd.read_csv` lines in the [data note, section "How to get it"](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Assessment/lfb-data.md). Keep `IncidentNumber` as text in both. Then profile your slice's columns. The questions are the same for every slice, and the [Demonstration, section "Part 1: The audit"](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Week/07/Demonstration.ipynb) runs them all:

- How many values are missing, and does a blank mean something? Use `.isna().sum()`
  - `SpecialServiceType` is blank on every fire and false alarm by design
  - `SecondPumpArriving_AttendanceTime` is blank where one engine was enough
  - A blank `FirstPumpArriving_AttendanceTime` is the data note's second challenge. Check what `NumPumpsAttending` shows on those rows
- Is every column the type it should be? Use `.info()` and `.describe()`
  - Every `DateAndTime*` column arrives as text and needs parsing
  - `PerformanceReporting` arrives as text because it mixes `1`, `2`, and `Not Used`
  - If every type is right, report that as a finding
- Are categories consistent? Use `.value_counts()` on your slice's categorical columns
  - Borough names should be in capitals in both tables. Check that they are
  - Station names appear in `IncidentStationGround`, `FirstPumpArriving_DeployedFromStation`, and `DeployedFromStation_Name`. An incident's station ground may differ from the station its engine came from
- Are there duplicated rows? Use `.duplicated().sum()` on each table and `mobilisations["ResourceMobilisationId"].duplicated().sum()`. The *what* slice runs this for everyone
- Do the tables join? Use `mobilisations["IncidentNumber"].isin(incidents["IncidentNumber"])`, counted both ways. The *when* slice runs this for everyone
  - This is the data note's first challenge. Expect around 1,400 engine rows whose incident is not in the table, and many incidents with no engine row at all
- What ranges look impossible? Use `.describe()` on numeric columns
  - Watch for an attendance time of one second, a `PumpMinutesRounded` in the tens of thousands, and a notional cost near half a million pounds. Each one is either a real major incident or a data problem

Ask your [Associate](https://github.com/REPPL/Teaching/blob/main/MN5813/2026-27/Guides/glossary.md#term-associate) to draft the profiling code. Check every line of it before you report the numbers it produces. Add that session to your Prompt Diary too.

**Then, in turn (about 10 minutes, one to two minutes each):** Each member **types a short findings summary into your team's record**. Write three to five bullet points with concrete numbers. For example:

- "`FirstPumpArriving_AttendanceTime` blank in 5.4% of incidents, and `NumPumpsAttending` is 1 on most of those"
- "1,415 engine rows carry an incident number the incidents table does not have"

Besides the challenges listed in the data note, expect the reports to include at least these:

- `DateAndTimeReturned` blank for almost every engine, and in more than one format where it is filled
- `DelayCode_Description` is blank for about seven engines in ten, and it also carries `Not held up` as a value. A blank and `Not held up` may mean different things
- A couple of thousand repeated `ResourceMobilisationId`s
- `IncidentGroup` blank on a handful of incidents

Aim for entries like this one:

```
DECISION: left join mobilisations to incidents on IncidentNumber. Keep
the 1,415 engine rows whose incident is missing in a separate table.
The two extracts are cut on different days, which explains the gap, and
the station-workload question needs every engine. We rejected dropping
them. Owner: Sam writes the join cell and prints the row counts before
and after.
```

Log a position on each of these before you leave the room:

- **The join**: Inner or left, on `IncidentNumber`, with the row counts before and after typed into the log. Include what happens to the engines whose incident is not in the table
- **Empty attendance times**: Which incidents each response-time figure covers, what it excludes, and where the report states it
- **Duplicates**: What counts as a duplicate when one incident has several engine rows. Include what you do with a repeated `ResourceMobilisationId`
- **Timestamps**: Which `DateAndTime*` columns you parse, with which explicit format, and whether you parse `DateAndTimeReturned` at all
- **Names**: Which borough column and which station column the team uses when both tables carry one. Include the casing rule you apply before any join or group on a name
- **Scope**: The whole extract, or a stated window (complete months only, or a single calendar year). Log the choice either way

Log disagreements too. Two of you may keep arguing the same point.
